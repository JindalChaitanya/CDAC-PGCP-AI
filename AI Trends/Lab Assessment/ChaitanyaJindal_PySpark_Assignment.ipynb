```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum, countDistinct
```

```python
spark = SparkSession.builder \
    .appName("Transaction Analysis") \
    .getOrCreate()
```

```python
df = spark.read.csv(
    "txns.csv",
    header=True,
    inferSchema=True
)
```

```python
df.show(5)
```

```python
df.printSchema()
```

### Question 1
- Find the total number of transactions for each customer ID

```python
print("Total Number of Transactions for Each Customer")

transaction_count = df.groupBy("custid") \
                      .count() \
                      .orderBy("custid")

transaction_count.show()
```

### Question 2
- Find the total transaction amount of each customer ID

```python
print("Total Transaction Amount for Each Customer")

total_amount = df.groupBy("custid") \
                 .sum("amount") \
                 .withColumnRenamed("sum(amount)", "Total_Amount") \
                 .orderBy("custid")

total_amount.show()
```

### Question 3
- Find the total number of customers who spent using Credit Card

```python
credit_customers = df.filter(df.spendby == "credit") \
                     .select(countDistinct("custid")) \
                     .collect()[0][0]

print("Total Customers who spent by Credit Card :", credit_customers)
```

### Question 4
- Find the total number of customers who spent using Cash

```python
cash_customers = df.filter(df.spendby == "cash") \
                   .select(countDistinct("custid")) \
                   .collect()[0][0]

print("Total Customers who spent by Cash :", cash_customers)
```
