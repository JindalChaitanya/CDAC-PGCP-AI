### Problem Statement

Dataset: Demonetization_tweets.csv

Tasks:
1. Find the Top 50 Mentions (@)
2. Find the Top 50 Hashtags (#)
3. Find Tweets Mentioning the Prime Minister
4. Use Prepositions to Extract Relevant Information from Tweets

### Importing Libraries

```python
import nltk
import re

import pandas as pd

from collections import Counter
```

### Loading Dataset


```python
df = pd.read_csv("demonetization_tweets.csv", encoding='latin-1')

print("Number of Tweets:", len(df))
df.head()
```

### Identifying the Tweet Text Column


```python
print(df.columns)
```

```python
tweets = df['text'].astype(str)
```

### Task 1: Top 50 Mentions (@)

```python
mentions = []

for tweet in tweets:
    mentions.extend(re.findall(r'@\w+', tweet))

top_mentions = Counter(mentions).most_common(50)

top_mentions_df = pd.DataFrame(top_mentions,
                               columns=['Mention','Frequency'])

top_mentions_df
```

### Task 2: Top 50 Hashtags (#)


```python
hashtags = []

for tweet in tweets:
    hashtags.extend(re.findall(r'#\w+', tweet))

top_hashtags = Counter(hashtags).most_common(50)

top_hashtags_df = pd.DataFrame(top_hashtags,
                               columns=['Hashtag','Frequency'])

top_hashtags_df
```

### Task 3: Tweets Mentioning Prime Minister


```python
pm_tweets = tweets[
    tweets.str.contains(
        r'Prime Minister|PM|Modi|Narendra Modi',
        case=False,
        na=False
    )
]

print("Number of Tweets Found:", len(pm_tweets))
```

### Display First 20 Tweets Mentioning Prime Minister


```python
for i, tweet in enumerate(pm_tweets.head(20), start=1):
    print(f"\nTweet {i}:")
    print(tweet)
```

### Task 4: Extracting Prepositions from Tweets

```python
prepositions = []

for tweet in tweets:
    
    tokens = nltk.word_tokenize(tweet)
    pos_tags = nltk.pos_tag(tokens)
    
    for word, tag in pos_tags:
        if tag == 'IN':
            prepositions.append(word.lower())

preposition_freq = Counter(prepositions)

pd.DataFrame(
    preposition_freq.most_common(50),
    columns=['Preposition','Frequency']
)
```

### Sample Tweets with Prepositions

```python
for tweet in tweets.head(20):
    
    tokens = nltk.word_tokenize(tweet)
    pos_tags = nltk.pos_tag(tokens)
    
    extracted = [word for word, tag in pos_tags if tag == 'IN']
    
    if extracted:
        print("\nTweet:")
        print(tweet)
        print("Prepositions:", extracted)
```
