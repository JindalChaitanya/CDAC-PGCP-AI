### Importing Libraries

```python
import nltk
import spacy
from spacy import displacy
```

### Input Sentence

```python
text = "On the coming Tuesday, I will be attending a conference on Natural Language Processing in Berlin."
```

### Task 1: Tokenization using NLTK

```python
tokens = nltk.word_tokenize(text)

print("Tokens:")
print(tokens)
```

### Task 2: Part-of-Speech (POS) Tagging using NLTK

```python
pos_tags = nltk.pos_tag(tokens)

print("POS Tags:")
print(pos_tags)
```

### Task 3: Dependency Parsing using spaCy

```python
nlp = spacy.load("en_core_web_sm")

doc = nlp(text)

for token in doc:
    print(token.text, "->", token.dep_, "->", token.head.text)
```

### Task 4: Finding the Root Word


```python
for token in doc:
    if token.dep_ == "ROOT":
        print("Root Word:", token.text)
```

### Task 5: Visualizing the Dependency Tree

```python
displacy.render(doc, style="dep", jupyter=True)
```

### Conclusion

```python
print("Dependency Tree has been visualized using spaCy.")
print("Tokenization and POS Tagging have been performed using NLTK.")
print("The Root Word of the sentence is: attending")
```
