```python
import spacy
from spacy.matcher import Matcher
```

```python
text = "Developed countries such as USA have a better healthcare infrastructure compared to developing countries."
```

```python
nlp = spacy.load("en_core_web_lg")
```

```python
doc = nlp(text)
```

```python
for token in doc:
    print(f"{token.text} => {token.pos} => {token.dep_}")
```

```python
pattern = [
    {'POS':'NOUN'},
    {'LOWER':'such'},
    {'LOWER':'as'},
    {'POS':'PROPN'},
]
```

```python
matcher = Matcher(nlp.vocab)
```

```python
matcher.add("Pattern X such as Y", [pattern])
```

```python
matches = matcher(doc)
matches
```

```python
for match_id, start, end in matches:
    span = doc[start:end]
    print(f"Pattern Name: {nlp.vocab.strings[match_id]}")
    print(f"Match: {span.text}")
    print(f"Start: {start}")
    print(f"End: {end}")
```

```python
text = "I love eating healthy fruits such as Apple. I like supercars such as Lamborghini."
doc = nlp(text)
for token in doc:
    print(f"{token.text} => {token.pos} => {token.dep_}")
pattern = [
    {'POS':'NOUN'},
    {'LOWER':'and'},
    {'LOWER':'other'},
    {'POS':'PROPN'},
]
matcher = Matcher(nlp.vocab)
matcher.add("Pattern X such as Y", [pattern])
matches = matcher(doc)
print(matches)
for i, (match_id, start, end) in enumerate(matches):
    print(i)
    span = doc[start:end]
    print(f"Pattern Name: {nlp.vocab.strings[match_id]}")
    print(f"Match: {span.text}")
    print(f"Start: {start}")
    print(f"End: {end}")
```
