### Problem Statement

Read the file "moon.txt" and perform the following tasks:

1. Perform Whitespace Tokenization and display first 20 tokens.
2. Perform Regex Tokenization and compare output with Whitespace Tokenization.
3. Annotate the words with Part-of-Speech (POS) tags.
4. Count the number of nouns and verbs in the text.

### Importing Libraries

```python
import nltk

from nltk import pos_tag
from nltk.tokenize import WhitespaceTokenizer
from nltk.tokenize import RegexpTokenizer
from collections import Counter
```

### Reading moon.txt File


```python
with open("moon.txt", "r", encoding="utf-8") as file:
    text = file.read()

print(text[:500])
```

### Task 1: Whitespace Tokenization


```python
wt = WhitespaceTokenizer()

white_tokens = wt.tokenize(text)

print("Total Tokens:", len(white_tokens))
print("\nFirst 20 Tokens:")
print(white_tokens[:20])
```

### Task 2: Regex Tokenization

```python
rt = RegexpTokenizer(r'\w+')

regex_tokens = rt.tokenize(text)

print("Total Tokens:", len(regex_tokens))
print("\nFirst 20 Tokens:")
print(regex_tokens[:20])
```

### Comparing Whitespace and Regex Tokenization


```python
print("Whitespace Tokens (First 20):")
print(white_tokens[:20])

print("\nRegex Tokens (First 20):")
print(regex_tokens[:20])
```

### Observation


```python
print("Whitespace Tokenizer keeps punctuation attached to words.")
print("Regex Tokenizer removes punctuation and returns cleaner tokens.")
```

### POS Tagging


```python
pos_tags = pos_tag(regex_tokens)

print("First 50 POS Tagged Words:\n")

for word, tag in pos_tags[:50]:
    print(word, ":", tag)
```

### Counting Nouns and Verbs


```python
noun_tags = ['NN', 'NNS', 'NNP', 'NNPS']
verb_tags = ['VB', 'VBD', 'VBG', 'VBN', 'VBP', 'VBZ']

noun_count = 0
verb_count = 0

for word, tag in pos_tags:
    
    if tag in noun_tags:
        noun_count += 1
        
    if tag in verb_tags:
        verb_count += 1

print("Number of Nouns :", noun_count)
print("Number of Verbs :", verb_count)
```

### Frequency of POS Tags


```python
tag_counts = Counter(tag for word, tag in pos_tags)

print("Most Common POS Tags:\n")
print(tag_counts.most_common(10))
```

### Conclusion


```python
print("Whitespace Tokenization completed.")
print("Regex Tokenization completed.")
print("POS Tagging completed.")
print("Nouns and Verbs counted successfully.")
```
