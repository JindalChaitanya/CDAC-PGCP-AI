### 1. Strings – Consecutive Vowels
- Problem: Find the maximum number of consecutive vowels in a given string.
- Technique Used: Sliding Window / Single Pass
- Logic: Increase count for vowel, reset count for consonant.
- Time Complexity: O(n)
- Space Complexity: O(1)

```python
in_str = input("Enter String: ").lower()
max_count = 0
count = 0
for char in in_str:
    if char in "aeiou":
        count += 1
        max_count = max(max_count, count)
    else:
        count = 0
print(max_count)
```

### 2. Strings – RGB Signal Light Problem
- Problem: Predict final signal colors after K minutes.
- Transition Cycle: R → G → B → R (cycle length = 3).
- Optimization: Use K % 3 to avoid large iterations.
- Time Complexity: O(n)

```python
lights = ["R", "G", "B"]
input_cycle = int(input("Enter Cycle Length: "))
after_cycle = input_cycle % 3
print(lights[after_cycle])
```

### 3. Strings – Maximum Distance of 'a' and Minimum Distance of 'b'
- X = Maximum distance between any two 'a' characters.
- Y = Minimum distance between any two 'b' characters.
- Approach: Track indices of 'a' and 'b'.
- X = last_a_index − first_a_index.
- Y = minimum difference between consecutive b indices.
- Final Answer = X + Y.

```python
in_str = input("Enter String: ")
first_a = in_str.find("a")
last_a = in_str.rfind("a")
X = last_a - first_a
first_b = in_str.find("b")
next_b = in_str.find("b", first_b + 1)
Y = next_b - first_b
print(X + Y)
```
